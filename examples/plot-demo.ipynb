The last cell is a dependency-free SVG fallback that renders in any Python environment.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 4 * np.pi, 600)
y = np.sin(x) * np.exp(-0.12 * x)

plt.figure(figsize=(8, 3.6))
plt.plot(x, y, color="#2563eb", linewidth=2, label="damped sine")
plt.axhline(0, color="#9ca3af", linewidth=0.8)
plt.title("damped sine wave")
plt.xlabel("x")
plt.ylabel("amplitude")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
```

## No matplotlib? Use the SVG fallback

The next cell uses only the standard library and the bridge's `display()` helper.


```python
import math

W, H = 760, 280
points = []
for i in range(500):
    x = i / 499 * 4 * math.pi
    y = math.sin(x) * math.exp(-0.12 * x)
    points.append(f"{40 + i * (W - 80) / 499:.2f},{H / 2 - y * 90:.2f}")

svg = (
    f'<svg xmlns="http://www.w3.org/2000/svg" width="{W}" height="{H}" viewBox="0 0 {W} {H}">'
    f'<rect width="100%" height="100%" fill="white"/>'
    f'<line x1="40" y1="{H/2}" x2="{W-40}" y2="{H/2}" stroke="#d1d5db"/>'
    f'<polyline fill="none" stroke="#2563eb" stroke-width="2" points="{" ".join(points)}"/>'
    f'<text x="{W/2}" y="26" text-anchor="middle" font-family="sans-serif" font-size="16" fill="#111827">damped sine wave (SVG)</text>'
    '</svg>'
)

class SVG:
    def _repr_svg_(self):
        return svg

display(SVG())
```
